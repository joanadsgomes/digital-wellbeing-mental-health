# Parte 2 — Padrões de utilização digital

Esta parte estuda a distribuição das horas de ecrã, a dependência digital por género e região e as diferenças por tipo de dispositivo. A análise escrita foi mantida junto às visualizações correspondentes.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv("data/dataset_prog.csv", sep=";")
```

## KDE PLot - horas de ecrã por dia

```python
data = pd.read_csv("data/dataset_prog.csv", sep=";")

plt.figure(figsize=(7,5))
sns.kdeplot(data=data, x='device_hours_per_day', fill=True, linewidth=2)
plt.title("distribuicao da variavel: horas de ecra por dia")
plt.xlabel("horas de ecra por dia")
plt.ylabel("densidade")
plt.show()
```

A distribuição do tempo de horas de ecrã por dia é fortemente concentrada em torno de 6 a 9 horas, mas é caracterizada por uma cauda longa de utilizadores que passam significativamente mais tempo em frente aos ecrãs.

## Relação de género, região e idade com o número de horas de ecrã

```python
data = pd.read_csv("data/dataset_prog.csv", sep=";")

sns.displot(
    data=data,
    x="digital_dependence_score",
    hue="gender",
    col="region",
    col_wrap=3,
    bins=15
)

plt.show()
```

```python
media_dependencia_regiao_genero = (
    data
    .groupby(['region', 'gender'])['digital_dependence_score']
    .mean()
    .reset_index()
    .sort_values(by='digital_dependence_score', ascending=False)
)

print('Média do score de dependência digital por região e género: ') 
media_dependencia_regiao_genero
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=media_dependencia_regiao_genero,
    x='region',
    y='digital_dependence_score',
    hue='gender'
)

plt.title("Média da Dependência Digital por Região e Género")
plt.xlabel("Região")
plt.ylabel("Score Médio de Dependência Digital")
plt.xticks(rotation=25)
plt.show()
```

A análise da dependência digital em função da idade, género e região foi realizada através de um conjunto de seis histogramas, que correspondem às diferentes regiões em estudo, com distinção por género e com a distribuição da variável digital_dependence_score. Esta abordagem permite observar simultaneamente padrões de dependência digital em diferentes contextos geográficos e sociodemográficos. 
Como é possível observar, verificamos que a dependência digital não se distribui de forma homogénea entre as regiões, existindo diferenças visíveis na concentração dos valores, sendo estes por norma, mais elevados em grupos mais jovens (dos 18 aos 30 anos).
Para uma análise mais precisa, elaborámos uma média do score por região e género. Com isto, foi possível constatar que a maior média de dependência digital é observada na América do Sul, no género feminino com um valor de, aproximadamente, 38.65, seguida pela América do Norte, igualmente pelo género feminino, com um score de 37.37 e pela Ásia, por parte do género masculino com uma média de 37.32. Valores igualmente elevados surgem no continente africano, por parte do género feminino com uma média de 37.00 e no Médio Oriente, por parte do género masculino 36.92. Por outro lado, as médias mais baixas são registadas no continente europeu, por parte do género feminino, assumindo um valor de 35.90 e no continente sul americano, por parte do género masculino com um valor de 35.30.
Estes resultados confirmam que, o género feminino apresenta, geralmente, valores médios de dependência digital mais elevados em várias regiões. Para além disto, é possível constatar que existem diferenças regionais claras, sendo a América do Sul, a América do Norte e Ásia as regiões com valores médios mais elevados. Enquanto a Europa apresenta, os níveis médios mais baixos de dependência digital.
A relevância destes resultados torna-se ainda mais evidente quando articulada com o clustermap das correlações, no qual a variável digital_dependence_score surge fortemente associada a variáveis psicológicas, tais como ansiedade, depressão, stress e negativamente associado à qualidade do sono, felicidade e foco.
Deste modo, regiões e grupos com maiores médias de dependência digital correspondem aos grupos potencialmente mais vulneráveis em termos de saúde mental, validando empiricamente a relação entre comportamento digital excessivo e bem-estar psicológico. 


## Boxplot- tempo de ecrã por tipo de dispositvo

```python
data = pd.read_csv("data/dataset_prog.csv", sep=";")

plt.figure(figsize = (8,5))
sns.boxplot(data = data, x='device_type', y='device_hours_per_day')
plt.title('Horas de ecrã por tipo de dispositivo')
plt.show()
```

Tecnicamente, o boxplot é o gráfico mais adequado para este tipo e análise, uma vez que permite não só observar a tendência central, em relação ao número de horas de uso com recurso à mediana, mas também a dispersão dos dados e a presença de valores extremos. Como podemos observar, em relação ao tempo diário de ecrã, podemos constatar que este possui uma distribuição semelhante entre os diferentes tipos de dispositivos, com medianas próximas e níveis de dispersão comparáveis.
No entanto, também é importante denotar a presença consistente de outliers, pois estes possuem uma particular relevância, dado que representam os utilizadores com padrões de consumo excessivo. Estes, estão tendencialmente associados a maiores níveis de stress, ansiedade e perturbações de sono.
