# Parte 3 — Hábitos digitais e bem-estar

Analisamos associações entre horas de ecrã e stress, as distribuições por papel quotidiano (*daily role*) e a relação entre desbloqueios e notificações. Estes gráficos apresentam associações e não demonstram causalidade.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv("data/dataset_prog.csv", sep=";")
```

## Scatterplot - horas de ecrã vs stress

```python
data = pd.read_csv("data/dataset_prog.csv", sep=";")

plt.figure(figsize=(7,5))
sns.scatterplot(data=data, x='device_hours_per_day', y='stress_level', alpha=0.5)
sns.regplot(data=data, x='device_hours_per_day', y='stress_level', scatter=False, color='red')
plt.title("stress vs horas de ecra")
plt.show()
```

O scatterplot entre o número de horas de ecrã por dia e o nível de stress mostra uma grande dispersão dos pontos. A linha de regressão sobe ligeiramente, indicando que níveis mais elevados de horas de ecrã tendem a estar associados a um aumento do stress. A dispersão observada mostra que a relação não é forte, pelo que o tempo de ecrã, não explica, por si só, o nível de stress na sua totalidade.

## Violinplot - tempo de ecrã por daily role

```python
data = pd.read_csv("data/dataset_prog.csv", sep=";")

plt.figure(figsize=(10,6))
sns.violinplot(
    x='daily_role',
    y='device_hours_per_day',
    hue='daily_role',
    data=data,
    inner='quartile',
    palette="Set2",
    legend=False
)
plt.title("Distribuicao do Tempo de Ecra por Daily Role")
plt.xlabel("Daily Role")
plt.ylabel("Horas de Ecra por Dia")
plt.xticks(rotation=45)
plt.show()
```

As variáveis associadas ao estilo de vida, como o “daily_role”, influenciam diretamente a relação entre o uso digital e o bem-estar mental. Diferentes grupos apresentam níveis distintos de tempo de ecrã e utilizam os dispositivos por motivos variados. Por exemplo, estudantes e pessoas desempregadas tendem a apresentar tempos de ecrã mais elevados, muitas vezes associados a stress académico, procura de emprego ou maior exposição a conteúdos digitais, o que pode intensificar sintomas de ansiedade. Já trabalhadores a tempo inteiro ou parcial apresentam padrões mais estáveis de utilização, geralmente com menor impacto negativo. Assim, o efeito do uso digital no bem-estar mental não depende apenas da quantidade de horas, mas do contexto diário em que esse uso ocorre.

## Scaterplot - relação entre unlocks e notificações

```python
plt.figure(figsize = (7,5))
sns.scatterplot(data = data, x ='phone_unlocks', y= 'notifications_per_day', alpha= 0.5)
sns.regplot(data = data, x= 'phone_unlocks', y='notifications_per_day', scatter = False, color ='red')
plt.title('Unlocks vs. Notificações')

plt.show()
```

No presente gráfico, é possível identificar uma relação fraca positiva entre o número de desbloqueios do telemóvel e o número de notificações recebidas por dia. Apesar de existir uma tendência crescente, a elevada dispersão dos dados demonstra que o comportamento digital varia significativamente entre indivíduos.  A presença de utilizadores com valores extremamente elevados de notificações e desbloqueios sugere padrões de uso intensivo, potencialmente associados a maiores níveis de stress, ansiedade e dependência digital.
